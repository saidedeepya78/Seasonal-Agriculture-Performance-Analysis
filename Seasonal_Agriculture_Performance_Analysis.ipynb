**7. Missing Values in Dataset**

```python
# step 1: counting missing values in every column
missing_count = df.isnull().sum()

# step 2: calculating percentage of missing values for each column
missing_percent = (df.isnull().sum() / len(df)) * 100

# combining count and percentage into one table
missing_summary = pd.DataFrame({
    'Missing_Count': missing_count,
    'Missing_Percent': missing_percent
})

# showing only the columns that actually have missing values
missing_summary = missing_summary[missing_summary['Missing_Count'] > 0]

if missing_summary.empty:
    print("No missing values found in the dataset. No treatment was required.")
else:
    print("Columns with missing values:")
    print(missing_summary)
```

```python
# step 3: creating a copy so the original df is not touched
df_clean = df.copy()

# step 4: handling missing values only if any exist
if not missing_summary.empty:
    for col in missing_summary.index:

        # if more than half the column is missing, we flag it instead of filling blindly
        if missing_summary.loc[col, 'Missing_Percent'] > 50:
            print(f"Warning: '{col}' has more than 50% missing values. Review manually before filling.")
            continue

        # for numeric columns, fill missing values with the median
        if df_clean[col].dtype in ['int64', 'float64']:
            median_value = df_clean[col].median()
            df_clean[col] = df_clean[col].fillna(median_value)
            print(f"'{col}' (numeric) missing values filled with median: {median_value}")

        # for categorical/text columns, fill missing values with the mode (most common value)
        else:
            mode_value = df_clean[col].mode()[0]
            df_clean[col] = df_clean[col].fillna(mode_value)
            print(f"'{col}' (categorical) missing values filled with mode: {mode_value}")
else:
    print("df_clean created as a copy of df. No missing value treatment was needed.")
```

```python
# step 5: confirming missing values are handled
print("Remaining missing values in df_clean:")
print(df_clean.isnull().sum().sum())

# step 6: checking how many rows remain after cleaning
print("\nNumber of rows in original df:", df.shape[0])
print("Number of rows in df_clean:", df_clean.shape[0])
```

**10. Outlier Analysis**

```python
# step 1: automatically picking numeric columns for outlier analysis
numeric_cols = df_clean.select_dtypes(include=['int64', 'float64']).columns
print("Numeric columns being checked for outliers:")
print(list(numeric_cols))
```

```python
# step 2: calculating Q1, Q3, IQR and counting outliers for each numeric column

outlier_summary = []

for col in numeric_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # counting values that fall outside the lower and upper bounds
    outliers = df_clean[(df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)]
    outlier_count = outliers.shape[0]

    outlier_summary.append({
        'Column': col,
        'Q1': round(Q1, 2),
        'Q3': round(Q3, 2),
        'IQR': round(IQR, 2),
        'Lower_Bound': round(lower_bound, 2),
        'Upper_Bound': round(upper_bound, 2),
        'Outlier_Count': outlier_count
    })

# step 3: displaying results in a readable table
outlier_table = pd.DataFrame(outlier_summary)
print("Outlier summary for each numerical column:")
outlier_table
```

**14. Correlation Analysis**

```python
# step 1: correlation matrix for all numeric columns
numeric_cols = df_clean.select_dtypes(include=['int64', 'float64']).columns
corr_matrix = df_clean[numeric_cols].corr()

print("Correlation matrix:")
corr_matrix
```

```python
# step 2: heatmap visualization of the correlation matrix

plt.figure(figsize=(14, 10))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap - Numerical Variables')
plt.show()
```

```python
# step 3: converting the correlation matrix into a simple pair-wise table
# (each pair listed once, excluding a variable's correlation with itself)

corr_pairs = corr_matrix.unstack()
corr_pairs = corr_pairs[corr_pairs != 1.0]  # removing self-correlation (always 1.0)

# removing duplicate pairs (A-B and B-A are the same correlation)
corr_pairs = corr_pairs.reset_index()
corr_pairs.columns = ['Variable_1', 'Variable_2', 'Correlation']
corr_pairs['sorted_pair'] = corr_pairs.apply(lambda row: tuple(sorted([row['Variable_1'], row['Variable_2']])), axis=1)
corr_pairs = corr_pairs.drop_duplicates(subset='sorted_pair').drop(columns='sorted_pair')

# sorting by correlation strength (highest absolute value first)
corr_pairs['Abs_Correlation'] = corr_pairs['Correlation'].abs()
corr_pairs_sorted = corr_pairs.sort_values(by='Abs_Correlation', ascending=False).drop(columns='Abs_Correlation')

print("All variable pairs sorted by correlation strength:")
corr_pairs_sorted.reset_index(drop=True)
```

```python
# step 4: identifying the strongest positive and strongest negative correlation

strongest_positive = corr_pairs_sorted[corr_pairs_sorted['Correlation'] > 0].iloc[0]
strongest_negative = corr_pairs_sorted[corr_pairs_sorted['Correlation'] < 0].iloc[0]

print("Strongest Positive Correlation:")
print(f"  {strongest_positive['Variable_1']} & {strongest_positive['Variable_2']} -> {strongest_positive['Correlation']:.3f}")

print("\nStrongest Negative Correlation:")
print(f"  {strongest_negative['Variable_1']} & {strongest_negative['Variable_2']} -> {strongest_negative['Correlation']:.3f}")
```

```python
# step 5: table of "highly correlated" pairs (using a common threshold of 0.5)
# this threshold is just a common rule of thumb for "moderate to strong" correlation

threshold = 0.5
highly_correlated = corr_pairs_sorted[corr_pairs_sorted['Correlation'].abs() >= threshold]

if highly_correlated.empty:
    print(f"No variable pairs found with correlation strength >= {threshold}.")
else:
    print(f"Variable pairs with correlation strength >= {threshold}:")
    highly_correlated.reset_index(drop=True)
```

**25. Statistical Testing**

```python
# step 1: setting up - separating Yield values by season into groups

seasons = df_clean['Season'].unique()
print("Seasons found in dataset:", seasons)

yield_groups = [df_clean[df_clean['Season'] == season]['Yield_Tonnes_Ha'].dropna() for season in seasons]

for season, group in zip(seasons, yield_groups):
    print(f"{season}: {len(group)} records")
```

```python
# step 2: stating the hypotheses

print("Null Hypothesis (H0): There is no significant difference in average Yield_Tonnes_Ha across seasons.")
print("Alternative Hypothesis (H1): At least one season has a significantly different average Yield_Tonnes_Ha.")
print("Significance level (alpha): 0.05")
```

```python
# step 3: checking assumptions before choosing the test

# 3a. Normality check for each season's yield using Shapiro-Wilk test
print("Normality check (Shapiro-Wilk test) for Yield in each season:")
normal_flags = []

for season, group in zip(seasons, yield_groups):
    stat, p_value = stats.shapiro(group)
    is_normal = p_value > 0.05
    normal_flags.append(is_normal)
    print(f"  {season}: p-value = {p_value:.4f} -> {'Looks normal' if is_normal else 'Not normal'}")

# 3b. Equal variance check across all seasons using Levene's test
levene_stat, levene_p = stats.levene(*yield_groups)
equal_variance = levene_p > 0.05

print(f"\nLevene's test for equal variance: p-value = {levene_p:.4f} -> {'Variances look equal' if equal_variance else 'Variances are not equal'}")
```

```python
# step 4: choosing and running the appropriate test based on the assumption checks above

all_normal = all(normal_flags)

print("Assumption Summary:")
print(f"  All groups normally distributed: {all_normal}")
print(f"  Equal variance across groups: {equal_variance}")
print()

if all_normal and equal_variance:
    # assumptions reasonably met -> One-way ANOVA
    test_stat, p_value = stats.f_oneway(*yield_groups)
    test_used = "One-way ANOVA"
else:
    # assumptions not met -> Kruskal-Wallis (non-parametric alternative)
    test_stat, p_value = stats.kruskal(*yield_groups)
    test_used = "Kruskal-Wallis Test"

print(f"Test used: {test_used}")
print(f"Test statistic: {test_stat:.4f}")
print(f"P-value: {p_value:.6f}")
```

```python
# step 5: interpreting the result at alpha = 0.05

alpha = 0.05

print(f"Significance level (alpha): {alpha}")
print(f"P-value obtained: {p_value:.6f}")
print()

if p_value < alpha:
    print("Result: The p-value is less than alpha, so we REJECT the null hypothesis.")
    print("Conclusion: There IS a statistically significant difference in Yield across seasons.")
else:
    print("Result: The p-value is greater than or equal to alpha, so we FAIL TO REJECT the null hypothesis.")
    print("Conclusion: There is NOT enough statistical evidence of a difference in Yield across seasons.")
```

**28. Limitations**

1. No time/year dimension in the dataset
Looking at df_clean.columns, there is no Year or date-based column in this dataset — only Season (Kharif, Rabi, Zaid). This means the analysis captures seasonal patterns within a single snapshot rather than tracking how performance changes across multiple years. Any conclusions about "seasonal trends" reflect differences between seasons in this dataset only, not long-term trends over time.

2. Limited geographical coverage
The dataset covers only 8 states and 10 districts (with the same 10 district names repeating across every state). This is a small subset of India's agricultural regions, so findings about which states or districts perform better should not be generalized to states or regions not included in this dataset.

3. Dataset size varies unevenly across categories
While the overall dataset has 4000 records, this is unevenly split — for example, Zaid season has only 594–654 records compared to Kharif's ~1770–1780 and Rabi's ~1600–1627 (the exact counts differed slightly between missing-value handling and later cleaning). Sugarcane also has fewer records (305) than crops like Rice (690). Averages computed from smaller subgroups are naturally less stable than those from larger ones.

4. Missing values were present and had to be estimated
Three columns — Rainfall_mm (48 missing), Soil_Moisture_pct (40 missing), and Yield_Tonnes_Ha (32 missing) — had missing values that were filled using the median. While this is a reasonable approach, it means a small portion of the values used in this analysis are estimated rather than originally recorded, which could slightly affect precision in seasons or crops where those missing values were concentrated.

5. This is observational data, so no causation can be established
Every relationship explored in this project — Season vs Yield, Rainfall vs Yield, Water Efficiency vs Yield, Pesticide vs Disease Risk, and so on — is based on observing existing data, not on a controlled experiment. Even where correlations were found (such as the moderate Yield-Profit correlation of 0.490, or the strong Water Efficiency-Yield correlation of 0.915), this only shows that variables move together in this dataset. It does not prove that one factor directly causes another, since many unmeasured factors (soil quality, farmer experience, local practices) could be influencing both variables at once.

6. Outliers and skewed data were identified but not removed
The outlier analysis (using the IQR method) and the mean-vs-median comparison both showed that columns like Yield, Production, Water Used, and Profit are right-skewed, with Sugarcane's very high yield values likely contributing significantly to this. These records were kept in the dataset rather than removed, since they represent genuine data points, not necessarily errors — but this means average-based statistics in this project can be pulled upward by a relatively small number of high-value records, and should be read alongside the median for a fuller picture.

7. Statistical testing was limited to non-parametric comparison of one variable
The formal hypothesis test performed in this project (Kruskal-Wallis) only tested whether Yield differs significantly across Season, and was used specifically because the yield data did not meet the normality assumption required for a standard one-way ANOVA. Other variables (like Profit or Water Efficiency) and other groupings (like Crop or State) were explored descriptively and visually, but were not put through their own formal statistical significance tests in this project.
